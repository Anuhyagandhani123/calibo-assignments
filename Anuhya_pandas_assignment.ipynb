Q1. Array Operations and Vectorization

```python
import numpy as np
import pandas as pd

arr = np.array([10, 20, 30])
s = pd.Series([10, 20, 30])

print(arr * 2)
print(s * 2)
```

Similarity: Both perform element-wise multiplication by 2.


Difference: NumPy array displays only values, while a Pandas Series displays values along with an index.

Section C — Edge Cases

Q1. NumPy Broadcasting


Question:

Arrays have shapes (3,1) and (1,3).

```python
import numpy as np
a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

result = a + b

print(result)
print(result.shape)
```

NumPy broadcasting expands the (3,1) and (1,3) arrays to compatible (3,3) shapes.

If:

```python
b = np.array([1, 2])
print(b)
```

then the shapes (3,1) and (2,) are not compatible for broadcasting, so NumPy raises a ValueError.

Q2. CSV Parsing, Missing Values and Dates

Question:

The CSV contains "Asha, R", missing marks, and dates in different formats.


Answer:

The comma inside "Asha, R" is handled correctly because the name is enclosed in double quotes. Pandas treats it as one field.

Potential issues:

Dates have different formats.
Marks contains missing values.
"NA" may represent a missing value, but this should be confirmed from the dataset's meaning.

```python
df = pd.read_csv("students.csv")

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

df["Date"] = pd.to_datetime(
    df["Date"],
    errors="coerce",
    dayfirst=False
)

print(df)
print(df.dtypes)
```

errors="coerce" converts invalid values into NaN for numeric data and NaT for invalid dates.

"NA" should be investigated because it might mean a missing value, or it might have another meaning such as a special status.

Q1. Student Assessment Dataset

Question:
The dataset contains:

Missing marks
Duplicate records
IT, it, I.T.
Finance, finance
Marks stored as text
Absent
NA
88 marks
A mark of 105

Step 1 — Load and inspect

```python
import pandas as pd

df = pd.read_csv("student_assessment_dirty.csv")

print(df.head())
df.info()
print(df.isnull().sum())
```

Step 2 — Standardize departments

```python
df["Department"] = (
    df["Department"]
    .str.strip()
    .str.upper()
    .str.replace(".", "", regex=False)
)
print(df)
```

Step 3 — Remove duplicate records

```python
df = df.drop_duplicates()
print(df)
```

Step 4 — Convert Marks to numeric

```python
df["Marks"] = (
    df["Marks"]
    .astype("string")
    .str.extract(r"(\d+\.?\d*)", expand=False)
)

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")
print(df)
```

Step 5 — Investigate missing marks

```python
print(df[df["Marks"].isna()])
```

Step 6 — Identify invalid marks

```python
invalid_marks = df[(df["Marks"] < 0) | (df["Marks"] > 100)]

print(invalid_marks)
```

Step 7 — Display cleaned data

```python
print(df)
```
