Section A – Conceptual Understanding      



A1. NumPy arrays vs Python lists

NumPy arrays are preferred for numerical computation because they are faster and use memory more efficiently. They support vectorized operations, so calculations can be performed on many values at once without writing Python loops. NumPy also provides many mathematical and statistical functions useful in AI/ML.

A2. Vectorization

Vectorization means performing operations on an entire NumPy array at once instead of processing each element using a Python loop. It is important for large datasets because NumPy operations are optimized and generally execute much faster than explicit Python loops.

A3. A * B vs A @ B

A * B performs element-wise multiplication, where corresponding elements are multiplied. It can be used when applying element-wise operations to neural-network data.

A @ B performs matrix multiplication. It is commonly used in neural networks to multiply an input/feature matrix by a weight matrix.

A4. Broadcasting

Broadcasting allows NumPy to perform element-wise operations on arrays with different shapes. Starting from the last dimension, the dimensions are compatible when they are equal or when one of them is 1. Missing dimensions are treated as 1.

A5. flatten() vs ravel()

flatten() returns a flattened copy of the array, while ravel() usually returns a view when possible. A view can use less additional memory. Therefore, for memory-constrained large-image processing, ravel() is generally preferred when modifying the flattened result is not required.

A6. axis=0 and axis=1

For:

```python
import numpy as np

a = np.array([[1, 2, 3],
              [4, 5, 6]])
print(a)
```

axis=0 operates down the rows:

```python
np.sum(a, axis=0)
# [5 7 9]
```

axis=1 operates across the columns of each row:

```python
np.sum(a, axis=1)
# [ 6 15]
```

A7. Normalization vs standardization

Scaling features is important because features with very different numerical ranges can affect ML models unevenly. Normalization commonly converts values to a fixed range such as [0, 1]. Standardization transforms values so that they have approximately mean 0 and standard deviation 1.

A8. np.nanmean()

np.nanmean() calculates the mean while ignoring NaN values. If np.mean() is used on an array containing NaN, the result will normally be NaN.

A9. RGB image shape

For ML data, the convention is:

(samples, features)

For RGB images, the equivalent batch representation is commonly:

(samples, height, width, channels)

For 64 RGB images of size 128 × 128:

(64, 128, 128, 3)

Reproducibility allows the same experiment to produce the same random values when repeated. Using a fixed seed makes debugging and comparing experiments easier.

Section D – Applied Problems


D1. Data Preprocessing Pipeline

Steps

Detect missing values using np.isnan().
Calculate an appropriate replacement value, such as the mean.
Replace missing values using np.where().
Standardize the cleaned data.
Feed the resulting clean numerical data into the model.

```python
import numpy as np

data = np.array([
    [10.0, 20.0],
    [15.0, np.nan],
    [20.0, 30.0],
    [25.0, 35.0]
])

# Detect missing values
print("Missing values:")
print(np.isnan(data))

# Column means ignoring NaN
column_means = np.nanmean(data, axis=0)

# Replace NaN values
clean_data = np.where(
    np.isnan(data),
    column_means,
    data
)

# Standardize column-wise
mean = clean_data.mean(axis=0)
std = clean_data.std(axis=0)

standardized_data = (clean_data - mean) / std

print("Clean data:")
print(clean_data)

print("Standardized data:")
print(standardized_data)
```

D2. Image Batch Understanding

A single RGB image has the shape:

(height, width, channels)

For a 64 × 64 RGB image:

(64, 64, 3)

A batch of 16 RGB images has the shape:

(16, 64, 64, 3)


```python
import numpy as np

rng = np.random.default_rng(42)

images = rng.integers(
    0, 256,
    size=(16, 64, 64, 3)
)

print("Batch shape:", images.shape)

average_image = images.mean(axis=0)

print("Average image shape:", average_image.shape)
```

E1. Per-Column Min-Max Normalization

```python
import numpy as np

X = np.array([
    [10, 100, 1000],
    [20, 200, 2000],
    [30, 300, 3000]
])

column_min = X.min(axis=0)
column_max = X.max(axis=0)

normalized = (X - column_min) / (column_max - column_min)

print(normalized)
```
